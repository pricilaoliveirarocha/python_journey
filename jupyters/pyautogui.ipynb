```python
# %%
"""
Automação para download, leitura de uma planilha de vendas e envio de email com análise desses dados.

O script realiza as seguintes etapas:
1. Abre o navegador Brave.
2. Maximiza a janela ativa.
3. Acessa uma pasta específica do Google Drive.
4. Abre e baixa uma planilha.
5. Carrega o arquivo Excel utilizando pandas.
6. Exibe os dados da planilha.

Bibliotecas utilizadas:
- pyautogui: automação de teclado e mouse.
- pygetwindow: manipulação da janela ativa.
- pandas: leitura e manipulação da planilha.
- time: controle de espera entre ações.
"""

# %%
```

```python
# 

import pyautogui as ag
import pyperclip as pc
import time
import pygetwindow as gw

ag.PAUSE = 1 # Edita a configuração do autogui para pausar entre uma chamada e outra 1s
```

```python
link = 'https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga'

# Abre o navegador 
ag.press('win')
ag.write('brave')
ag.press('enter')

# Verifica se a janela está maximizada
janela = gw.getActiveWindow()
if janela is not None and not janela.isMaximized:
    janela.maximize()

# Acessa o link
ag.write(link)
ag.press('enter')
```

```python
time.sleep(3) # Espera o navegador abrir

# Clica no arquivo para abrir
ag.click(x=661, y=428, clicks=2) 

# Clica com o botão direito e baixa o excel
ag.click(x=661, y=428, clicks=1, button="secondary") 
time.sleep(2)
ag.click(x=795, y=498)
```

```python
# Importar a base de dados
import pandas as pd

caminho = r'C:\Users\Pricila Oliveira\Downloads\Vendas - Dez.xlsx'
tabela = pd.read_excel(caminho)

display(tabela)

# Contagens de dados
faturamento = tabela['Valor Final'].sum()
quantidade_produtos = tabela['Quantidade'].sum()

print(faturamento)
print(quantidade_produtos)
```
